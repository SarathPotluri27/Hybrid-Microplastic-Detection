```python
import os, shutil, random

source = "dataset_microplastics/micro_plastic"    # your original folder
target = "microplastic_yolo"

classes = os.listdir(source)
os.makedirs(f"{target}/train", exist_ok=True)
os.makedirs(f"{target}/val", exist_ok=True)

for cls in classes:
    imgs = os.listdir(f"{source}/{cls}")
    random.shuffle(imgs)

    split = int(0.8 * len(imgs))
    train_imgs = imgs[:split]
    val_imgs = imgs[split:]

    os.makedirs(f"{target}/train/{cls}", exist_ok=True)
    os.makedirs(f"{target}/val/{cls}", exist_ok=True)

    for img in train_imgs:
        shutil.copy(f"{source}/{cls}/{img}", f"{target}/train/{cls}/")

    for img in val_imgs:
        shutil.copy(f"{source}/{cls}/{img}", f"{target}/val/{cls}/")

print("Dataset prepared successfully!")
```

```python
from ultralytics import YOLO
from PIL import Image, ImageDraw
import os

# Load model
model = YOLO(r"C:\Users\indra\Downloads\projects-2025\micropl\runs\classify\train11\weights\best.pt")

# Input image
img_path = r"C:\Users\indra\Downloads\projects-2025\micropl\runs\detect\predictions\177_jpg.rf.78ce205aa324b8cb17ee4051188b13d5.jpg"

# Run prediction
results = model.predict(img_path)
pred = results[0]

# Load image for drawing
img = Image.open(img_path).convert("RGB")
draw = ImageDraw.Draw(img)

# Draw a rounded circle over the image center
w, h = img.size
radius = int(min(w, h) * 0.35)

center = (w//2, h//2)
bbox = [
    center[0] - radius,
    center[1] - radius,
    center[0] + radius,
    center[1] + radius
]

# Blue rounded highlight
draw.ellipse(bbox, outline="blue", width=6)

# Add predicted label text
label = pred.names[pred.probs.top1]
score = pred.probs.top1conf

draw.text((10, 10), f"{label}: {score:.2f}", fill="blue")

# Save the output
out_path = os.path.splitext(img_path)[0] + "_rounded.jpg"
img.save(out_path)

print("Saved:", out_path)
img.show()
```

```python
from ultralytics import YOLO
from PIL import Image, ImageDraw, ImageFont
import os

# Load model
model = YOLO(r"C:\Users\indra\Downloads\projects-2025\micropl\runs\classify\train11\weights\best.pt")

# Input image
img_path = r"C:\Users\indra\Downloads\projects-2025\micropl\runs\detect\predictions\177_jpg.rf.78ce205aa324b8cb17ee4051188b13d5.jpg"
results = model.predict(img_path)[0]  # single result

# Get the best prediction
probs = results.probs  # class probabilities
top_class_id = probs.top1
top_class_name = model.names[top_class_id]
confidence = float(probs.top1conf) * 100

print(f"PREDICTION → {top_class_name} ({confidence:.2f}%)")

# ------------------------------
# Draw prediction on image
# ------------------------------
img = Image.open(img_path).convert("RGB")
draw = ImageDraw.Draw(img)

# Font
try:
    font = ImageFont.truetype("arial.ttf", 32)
except:
    font = ImageFont.load_default()

text = f"{top_class_name} ({confidence:.2f}%)"

# Add black background rectangle
text_bbox = draw.textbbox((10, 10), text, font=font)
draw.rectangle(text_bbox, fill=(0, 0, 0, 160))

# Draw text
draw.text((10, 10), text, fill="white", font=font)

# Save result
out_path = "prediction_output.jpg"
img.save(out_path)
print("Saved prediction image at:", out_path)
```
